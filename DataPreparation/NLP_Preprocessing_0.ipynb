```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('IMDB-Dataset.csv')
df.head()
```

```python
# first we change this case into an lowercase
df['review'] = df['review'].str.lower()
df.head()
```

```python
df['review'][3]
```

```python
# Removing the HTML tags from the text using REGEX
import re
def remove_tags(text):
  pattern = re.compile('<.*?>')
  return pattern.sub(r'',text)

df['review'] = df['review'].apply(remove_tags)
df['review'][3]
```

```python
# Removing the Punctuations from the Text using the REGEX (!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~)
import string,time
string.punctuation

# slow method
def remove_punc(text):
  for char in string.punctuation:
    text = text.replace(char,'')
  return text

time1 = time.time()
df['review'] = df['review'].apply(remove_punc)
time2 = time.time() - time1
print(time2)


```

```python
# fast method
table = str.maketrans('','',string.punctuation)

def remove_punc1(text):
  return text.translate(table)

time3 = time.time()
df['review'] = df['review'].apply(remove_punc1)
time4 = time.time() - time3
print(time4)
```

```python
# fastest amoung this 3
# Create a regex pattern: '[!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~]'
punct_pattern = f"[{string.punctuation}]"

# Time the vectorized approach
time_start = time.time()
df['review'] = df['review'].str.replace(punct_pattern, '', regex=True)
time_end = time.time() - time_start

print("Vectorized Regex Time:", time_end)
```

```python
def chat_convertion(text):
  new_text = []
  for w in text.split():
    if w.upper() in short_forms:
      new_text.append(short_forms[w.upper()])
    else:
      new_text.append(w)
  return " ".join(new_text)
```
